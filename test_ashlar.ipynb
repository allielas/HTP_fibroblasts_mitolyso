````python
def make_fileseries_string(input_path, overlap=0.0, pixel_size=0.0949, width=5, height=4, layout="snake", direction="horizontal"):
    
    location = input_path.split("/")[8]
    block = input_path.split("/")[9][-1]

    pattern = "MAX_ch{channel:1}-"+location+"_block"+block+"_tile{series:2}.tif"
    
    fileseries_string = f"fileseries|{input_path}|pattern={pattern}|overlap={overlap}|pixel_size={pixel_size}|width={width}|height={height}|layout={layout}|direction={direction}"
    #print(fileseries_string)
    #construct command i.e. ashlar "fileseries|/mnt/bigdisk1/AllieSpangaro/Morphology_Replicative_Age_Project/MRC-5_MAX_SUM_PROJ/20250328_rep05/test_stitching/r02c01/Block_A|pattern=MAX_ch{channel}-r02c01_blockA_tile{series}.tif|overlap=0.1|pixel_size=0.0949|width=5|height=4|layout=snake|direction=horizontal"
    return fileseries_string

def make_output_filepath(input_path):
    output_path = os.path.join(input_path,"ashalr_img")
    os.makedirs(output_path, exist_ok=True)
    
    location = input_path.split("/")[8]
    block = input_path.split("/")[9][-1]
    
    output_file = os.path.join(output_path, location + "_block"+block+".ome.tif")
    #print(output_file)
    return output_file

#note: can also do ashlar Index.idx.xml --plates on raw data from Opera
````

## Now make the command

````python
def make_ashlar_command(input_path,registration_channel=0):
    fileseries_string = make_fileseries_string(input_path)
    output_file = make_output_filepath(input_path)
    
    ashlar_command = f"ashlar \"{fileseries_string}\" -c {registration_channel} -o {output_file}"
    #print(ashlar_command)
    return ashlar_command
    
    
````
